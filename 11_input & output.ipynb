#### ->This function accepts input from the keyword.

```python
name=input("Enter the name:")
print("You entered name:",name)
```

### 4.Convert from string to int:
#### ->We can convert the string value into int value.
#### ->int(p) is a predined function.
#### ->This function converts to int data type.

```python
a='123'
print(a)
print(type(a))
b=int(a)
print(b)
print(type(b))
```

```python
a=input("Enter the value:")
print("You entered value:",a)
print("data type is:",type(a))
```

### 5.Convert from string to other type:
#### ->we can convert the string value to float value.
#### ->float(p) is a predifined function.
#### ->This function converts to float data type.

```python
d="12.99"
print(d)
print(type(d))
b=float(d)
print(b)
print(type(b))
```

### 6.Convert float type to int type:
#### ->int(p) predifined function in python.
#### ->This function converts value into int data type.

```python
a=1000.45
print(a)
print(type(a))
b=int(a)
print(a)
print(type(a))
```
