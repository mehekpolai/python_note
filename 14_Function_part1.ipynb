#### b.Function with parameter or  parameterized function:
- If a function having parameter ,then the function is called  parameterized function.
- When we pass the parameters then,
   * Functions capture parameters value
   * These values performs the operation
   * Finally it brings the result.
#### syntax: 
#### def function name(p1,p2):
####     body of the function
#### function calling

```python
# one parameterized function:
def dance(a):
    print("hey Radhe let's",a)
dance("sing song")
```

```python
# two parameterized function:
def dance(a,b):
    print("hey Radhe let's",a)
    print("hey krishna let's",b)
dance("sing song","play flute")
```

#### 8. A function can return multiple values:
- In python , a function can return multiple values.
- Based on requrements a function can return multiple values.
  * If function is returning 2 values then while function calling we need to assign to 2 variables.
  *  If function is returning 3 values then while function calling we need to assign to 3 variables.
  *  If function is returning more thyan 1 values then while function calling we need to assign one variables then all values will be stored in tuple.

```python
def rahul(l,m):
    c=l+m
    return c
d=rahul(10,20)
print(d)
```
