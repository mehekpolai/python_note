#### 1.Dictionary data structure:
- If we want to represents group of individual object as a single entity then we should go for below data structure.
   * list
   * set
   * tuple
- If want to represent a group of object as key-value pairs then we should go for,
   * dict or dictionary
- We can create dict by using,
   * Curly bracesw {} symbol
   * dict() predifined function
- Dictionary data structure contains key-value pairs.
- key-value,
   * In dictionary key-value pairs called are separated by colon(:)symbol.
   * One key-value pair is called as item.
   * In dictionary every item is separated by comma (,) symbol.
   * In dictionary duplicats keys are not allowed.
   *  In dictionary duplicats values are may be allowed.
   *  A dictionary keys and values can store same type of elements(homogeneous).
   *  A dictionary keys and values can store different type of elements(heterogeneous).
- In dictionary insertion order is preserved means not fixed.
- Dictionary size will increase dynamically.
- Dictionary data strcture can not store the element in index order.
    * Indexing / slicing concept are not applicable.
* Note :
- dict is a predefined class in python.
- Once if we create dictionary object means internally object is creating for dict class.

* SUMMARY:
  1. create -> {} curly braces/flower braces{} or dict() function
  2. Store  -> Same data type or different data type
  3. insertion order -> not fixed
  4. Duplicate ->not allowed in keys and allowed in values
  5. nature -> Mutuable
  6. index/slice -> not allowed
  7. size   ->having dynamic nature means,it will increase
- If any data structure not following insertion order then it not having index/slicing.

#### 7.Update dictionary:
- We can update the key in dictionary .
##### Syntax: d[key]=value
##### case:1
- While updating the key in dictionary, if key is not available then a new key will be added at the end of the dictionary with specified value.
##### case:2
- While updating the key in dictionary, if key is already existing then old value will be replaced with new value.

```python
# Ex:
d={10:'bini',11:'tuni',13:'funi'}
print(d)
d[11]="tiwari"
print(d)
```

```python
d={10:'kiranya',11:'turvi',13:'funny'}
print(d)
d[11]="piwari"
print(d)
```

#### 8.Removing and deleting elements from dictionary:
- By using __del__ keyword, we can remove the keys.
- By using clear() method we can clear the objects in dictionary.

##### By using del keyword:
###### Syntax: del d[key]
- As per the syntax, it deletes entry associated with the specified key.
- If the key is not available, then we will get keyError.

```python
d={1:'horry',2:'doory',3:'orry'}
print(d)
del d[1]
print(d)
```

##### We can delete total dictionary object:
###### Syntax: del nameofthedictionary
- It can delete the total dictionary object.
- Once it deletes then we can't access the dictionary.
- It showes the error called Nameerror.

```python
d={1:'horry',2:'doory',3:'orry'}
print(d)
del d
print(d)
```

###### IMP points:
- As per object oriented principle,
   * If we want to access instance methods,then we should access by using object name.
- So, all dict methods we should access by using dict object .
###### IMP Methods:
- clear() method
- keys() method
- values() method
- items() method
##### i.clear() method:
- It is a method in dict class,we should access this method by using dictionary object.
- This method removes all entries in dictionary.
- After deletes all entries,it just keeps empty dictionary.

```python
d={101:"ganesh",102:"sila",103:"kiran"}
print(d)
d.clear()
print(d)
```

##### ii.keys() method:
- It is a method in dict class,we should access this method by using dictionary object.
- This method returns all keys associated with dictionary.

```python
d={101:"ganesh",102:"sila",103:"kiran"}
print(d)
print(d.keys())
```

##### iii.values() method:
- It is a method in dict class,we should access this method by using dictionary object.
- This method returns all values associated with dictionary.

```python
d={101:"ganesh",102:"sila",103:"kiran"}
print(d)
print(d.values())
```

##### iv.items() method:
- It is a method in dict class,we should access this method by using dictionary object.
- This method returns all keys & values associated with dictionary.

```python
d={101:"ganesh",102:"sila",103:"kiran"}
print(d)
for k,v in d.items():
    print(k,v)
```

#### 11.Dictionary comprehension:
- A dictionary comprehension requests creating new diictionary from iterable object like a set,tuple,list,dictionary and range.
- Dictionary comprehension codes very concise way.

```python
squares={a:a*a for a in range(1,6)}
print(squares)
```

```python
##### synatx:
LHS = RHS
var = {part1 part2 part3}
part1= variable/expression(in key value)
part2= for loop
part3=condition(optional)
var={variable    for loop    condition}
```

```python
names=["ram","don","suru"]
result={name:name.upper() for name in names}
print(result)
```

```python
names=["kris","krushh","kriya"]
result={name:len(name) for name in names}
print(result)
```
