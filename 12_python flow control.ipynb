## 5. Looping:
- If we want to execute a group of statements in multiple times,then we should go for looping kind of execution.
  #### a.for loop
  #### b.while loop
#### a.for loop:
- for is a keywords in python.
- Basically ,for loop is used to get or iterate elements one by one from sequence like string,list,tuple e.t.c.
- While iterating elements from sequ8ence we can perform operation on every element.
#### syntax:
#### for variable in sequence:
#### statements

```python
#ex:
values=[10,20,30,"geeta"]
for value in values:
    print(value)
```

```python
#Ex:
values=[10,20,30,40]
for value in values:
    print(value+2)
```

#### b.while loop:
- while is a keyword in python.
- If you want to execute a group of statements repeatedly until the condition reaches to false,then we should  go for while loop.
#### Syntax:
- initialization
-      while condition:
-         while block statements
-            increment/decrement
- while loop contains on expression/condition.
- As per the syntax colon(:) is mandotory otherwise iut through syntax error.
- After while loop we need to follow indentation otherwise it throws indentationError.
- Condition gives the result as bool type,means either __True__ or __False__.
- If the condition result is True ,then while loop executes till condition reaches to false.
- If the condition results is false ,then while loop execution terminates.
#### Conclusion:
- Till condition results the True the while loop statement will be executed.
- If the condition reaches to False ;then while loop terminates to execution.

```python
#Ex:
x=1
while x<=5:
    print(x)
    x=x+1
print("End")
```

### 6.Break Statement:
- break is a keyword in python.
- The break statement can be used inside the loops.
- By using break we can break execution  based on some condition.
- Generally,break statement is used to terminal for and while loops.

```python
# Ex:
x=0
while x<=10:
    print("x=",x)
    x=x+1
print("out of loop")
```

```python
# Ex:
x=1
while x<=10:
    print("x=",x)
    x=x+1
    if x==5:
        break
print("out of loop")
```

### 8.Pass Statement:
- pass is keyword in python.
- The pass statement is used as a placeholder for future code.
- It is useful as a placeholder when a statement is required syntactically but no code needs to be executed.
- pass is a null operation ,when it is executed nothing happens.
- we can define an empty function class,method with pass statement.
