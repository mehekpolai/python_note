### 8.Membership operator:
#### Def 1:
-We can checki,if string or character  is a member of string or not by using __in__ and __not__ in operator.
#### Def 2:
- We can check,if string is substring of main string or not by using __in__ and __not__ in operator.
#### in operator:
- __in__ operator returns True,if the string or character found in the main string.

```python
a="Python"
print('p' in a)
print('x' in a)
```

#### not in operator:
- __not in__ operator returns positive result of in operator.
-  __in__ operator returns True,if the string or character not found in the main string.

```python
x="frong"
print('p' not in x)
print('g' not in x)
```

#### e.replace(p1,p2):
- replace(p1,p2) is a predifined method in str class.
- This method we should access by using string object.
- We can replace old string with new string by using replace(p1,p2) method.

```python
w1="java programming langauge"
print(w1)
w2=w1.replace("java","python")
print(w2)
```

#### replace method returns new string object:
- As we know string is immutable,so replace(p1,p2) method never perform changes on the existing string object.
- replace(p1,p2) method creates new string object, we can check this by using id(P) function.

```python
w1="java programming langauge"
w2=w1.replace("java","python")
print(w2)
print(w1)
print(id(w1))
print(id(w2))
```

#### Q.string is immutable,is replace(p1,p2) method will modify the string object?
- Once we create a string object,we can't change or modify the existing string object.
- This behaviour is called as immutuablity.
- If we are trying to change or modify the existing string object ,then with those changes a new string object will be created.
- So,replace(p1,p2) method will create new string object with the midification.
