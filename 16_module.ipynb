#### 1.Modules:
- In python a module means a saved python file.
- This file can contain a group of class, method,function and variables.
- Every python file means __.py__ or __.python__ extension file is called as a module.
#### additionmultiplication.py
- x=10 
- def addition(a,b):
-    print("sum of two values:",(a+b))
- def multipilcation(a,b):
-   print("multipilcation of two values:",(a*b))
* Now additionmultiplication.py file is a module.
* additionmultiplication.py module contains one variable and two functions.
#### 2.Import Module:
- import is a keyword in python.
- By using import keyword we can import (get) modules in our program.
- Once we imported the module then we can use members(variables,function & etc) of module.
#### ex: importing additionmultiplication module and calling members.

```python
import additionmultiplication
print(additionmultiplication.x)
additionmultiplication.addition(1,2)
additionmultiplication.multiplication(2,3)
```

* Make a note:
- Whenever we are using a module in our program,for that module file will be generated and stored in the harddisk permanently.
#### 3.Renaming or aliasing a module :
- __as__ is  a keyword in python.
- By using __as__ keyword we can rename/alias existing module.
#### Syntax: 
#### import modulename as aliasname
#### ex:
#### import additionmultiplication as admule
- Here additionmultiplication is module name admul is aliasname.
- We can access members by using alias name admul.
- Ex: Importing module:

```python
import additionmultiplication as admule
print(admule.x)
admul.addition(1,2)
admul.multiplication(2,3)
```

#### 4.from and import keyword:
- __from__ is keyword in python.
- We can import some specific members of module using __from__ keyword.
- The main advantage of __from__ keyword is we can acess member directly without using module name.
- Ex: from and import keyword

```python
from additionmultiplication import x
print(x)


output:
10
sum of two values:30
```

```python
# nameerror :name "multiplication" is not defined .
from additionmultiplication import x
print(x)
multiplication(10,20)


output:
10
NameError
```

#### 5.import*(Star symbol):
- We can use __*(Star symbol)__ to import all member of a module.
- We can import all members a module as using import *(Star symbol).

```python
# Ex: importing by *(Star symbol)
from additionmultiplication import*
print(x)
addition(10,20)
multiplication(10,20)

output:
10
sum of two values:30
multiplication of two values:200
```

#### 6.member aliasing:
- We can give alias name to the members of a module.
- ex: member alising:

```python
from additionmultiplication import x as y addition as add
print(y)
add(10,20)

output:
10
sum of two values:30
```
