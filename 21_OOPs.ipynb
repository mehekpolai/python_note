```python
class NameOfTheClass:
    def __init__(self):
        body of the constructure
```

##### i.What is the main pupose of the constructor?
- The main purpose of costructor is to initialize instance variables.

#### 9.Instance Variable:
##### i. What is instance variable?
- If the value of variable is changing from object to object such type of variable is called as instance variables.

##### ii.Separated copy instance variable for every object:
- For every object a separate copy of instance variable will be create.


```python
class Student:
    def __init__(self,n,m):
        self.n=n
        self.m=m
s1=Student("Ram",101)
s2=Student("Raj",102)
print("Std1 info:")
print("name:",s1.n)
print("Id:",s1.m)
print("Std2 info:")
print("name:",s2.n)
print("Id:",s2.m)
```

```python
class Scolar:
    def __init__(self,n,m):
        self.n=n
        self.m=m
    def details(self):
        print("Instance Method:")
s1=Scolar("Ram",101)
s1.details()
print(s1.n)
print(s1.m)
```

###### here,
- s1.details() --->calling instance method     syntax:object.method()
- print(s1.n)  --->calling instance variable   syntax:object.variable()
- print(s1.m)

##### iii.Declaring & accessing instance variable:
- We can declare instance variables inside consttructor.
- We can access instance variables by using object name.

```python
class Employee:
    def __init__(self):
        self.eno=10
        self.ename="lion"
        self.esal=10000
emp=Employee()
print("Eno:",emp.eno)
print("Ename:",emp.ename)
print("Esal:",emp.esal)
```

#### 10.Instance method:
- There are methods which acts upon instance variables of the class.
- This methods are bound with instances or objects,that's why called instance method.
- The first parameter for instance methods is self variable.
- Along with self variable it can contain other variable as well.


```python
class Demo:
    def __init__(self,b):
        self.b=b
    def m(self):
        print(self.b)
d=Demo(10)
d.m()
```
