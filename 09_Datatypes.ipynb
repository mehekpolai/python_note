## 4.Creating range of values:
#### -> we can create range of values by using range(p) predefined function.
#### a.range(p) function :
#### -> As discussed ,we can create range of values by using range(p) function.
#### ->Here 'p' should be integer,otherwise we will gwt error.
#### ex-

```python
a= range(10)
for b in a :
    print(b)
```

#### a.range(p) function :
#### ->As disscussed we can create range of values by using range(start,end) function.

## 5.Accessing range values by using for loop :
#### -> we can access range values by using for loop.
#### ex-

```python
r=range(10,15)
for value in r :
    print(value)
```
