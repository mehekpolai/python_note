## 1.Operator
#### ->An operator is symbol that performs an operation.
#### ->An operator acts on some variables, those variable are called as operand.
#### ->If an operator acts on single  o-perand then it's called as unary operator.
#### ->If an operator acts on 2 operand, then it's called binary operator.
#### ->I(f an operator acts on 3 operand ,then it's called as ternary operator.
#### ex-

```python
a=10
b=20
c=a+b
print(c)
```

#### Here a,b,c are called as operands,"+" symbol is called as operator.
