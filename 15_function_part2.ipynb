#### 1.Formal and actual argument:
- When a function is defined it may have some parameters.
- These parameters receives the values.
  * Parameters are called as fomal arguments.
  * When we call the function we should pass values or data to the function.
  * These values are called as actual arguments.

```python
def add(a,b):
    c=a+b
    print(c)
# call the function
x=10
y=15
add(x,y)
```

- a,b are called as fomal arguments.
- x,y are called as actual arguments.

#### d.variable length argument(*args):  
- *args also called as variable length arguments.
- sometimes our requirement can be like,need to provide more values to functionparameter to process the result.Here we can use *args concept.
- It represent single * symbol before the argument name.
- Internally the provided values will be stored in tuple.
#### Few points:
- If we define one parameterized function then during function calling we need to pass one value.
- If we define two parameterized function then during function calling we need to pass two value,if we pass more than two values then we will get error.
#### Syntax:
#### def name_of_the_function(*x):
#### body of the function
- *x is variable length argument.
- We can pass any number of values to this *x.
- Internally the provided values will be represented in tuple.

```python
def m(x):
    print(x)
m(15)
```

```python
def m(x):
    print(x)
m(15,45)
```

```python
def m(*x):
    print(x)
m(15)
```

```python
def m(*x):
    print(x)
m(15,45,75)
```

```python
def m(*x):
    print(x,sum(x))
m(15,100,200)
```

```python
def m(*x):
    print(x[4])
m(15,56,700,159,300,450)
```

#### 3.Anonymous function or lambda function:
- Generally we can create normal function by using __def__ function.
- __lambda__ is a keyword in python.
- By using lambda keyword we can create __lambda function__ .
- __lambda function__ will proceed the i/p and return the result.
- A function without name is called anonymous function.
#### Syntax:
#### lambda argument_list:expression

```python
c=lambda x,y,z:x+y+z
res=c(12,45,60)
print(res)
```

#### Advantages:
- By using lambada function .we can write very comcise code so that readability of the program will be improved.

```python
s=lambda x:x*x
d=s(4)
print(d)
```

- Here because of lambda keyword it creates Anonymous function.

#### Where Lambda functions fits exactly?
- Sometimes we can pass function as argument to another function.In such cases lamb da function are best choice.
- We can use lambda fuction very commonly with filter(),map(),and reduce() functions,these function expect function as argument.
#### Lambda function:
- a.map(p1,p2)function
- b.filter(p1,p2)function
- c.reduce(p1,p2)function/ sum()
#### a.map(p1,p2)function:
- map(function ,iterable) is a predifined function in python.
- This function takes iterable abd apply logic on every item and returns new iterable.
#### syntax:
#### map(function,sequence)

```python
a=[100,200,300,400,500]
b=map(lambda x:x+10,a)
x=list(b)
print(a)
print(x)
```

```python
without_gst=[100,200,300,400,500]
with_gst=map(lambda x:x+10,without_gst)
x=list(with_gst)
print(x)
```

#### b.filter(p1,p2)function:
- It is a predifined function in python.
- This function takes iterable and apply filtering logic on every item and returns new iterable.
#### syntax:
#### filter(function,sequence)

```python
a=[1,2,3,4,5,6,7,8,9]
b=filter(lambda x:x>6,a)
y=list(b)
print(y)
```

```python
item_cost=[1023,2000,300,412,512,60,7000,80000,9000]
gst_cost=filter(lambda x:x>3000,item_cost)
y=list(item_cost)
print(y)
```

#### c.reduce(p1,p2)function:
- It is a predifined function in funtool module.
- This function apply a function of 2 arguments cumulatively,from left to right.
- Finally this function reduce the sequence to a single value.
#### syntax:
#### reduce(function,sequence)

```python
from functools import reduce
item =[100,200,300,800]
total= reduce(lambda x,y:x+y,item)
print(total)
```

- For example ,reduce(lambda x,y:x+y,[1,2,3,4,5] calculates ((((1+2)+3)+4)+5)
