#### 1.Set data structure:
- We can create set data structure by using,
    * curly braces/flower braces{} symbol
    * set() function predefined
- A set can store group of object or elements.
    * A set can store same(homogeneous) type element.
    * A set can store different(heterogeneous) type of element.
- A set size will dynamically increase.
-In set insertion order is not preserved means not fixed.
   * If we insert element into 10,20,30 then there is no  guarentee for output as 10,20,30 . 
- Example:
    * input =>[10,20,30]
    * output =>[30,10,20]
- Duplicate elements are not allowed.
- Set having mutuable nature.
  * Mutual means once we create a list object then we can change or modify the content of set object.
- Set data structure can't store elements by using index oredr.
##### make a note:
- Inside set every object can be separated by comma(,) separator.
- set is a predifined class in python.
- Once if we create tuple object means inernally object is creating for set class.

##### SUMMARY:
 1. create -> {} curly braces/flower braces{} or set() function
  2. Store  -> Same data type or different data type
  3. insertion order -> not fixed
  4. Duplicate ->not allowed
  5. nature -> Mutuable
  6. index/slice -> not allowed
  7. sizee   ->having dynamic nature means,it will increase
- If any data structure not following insertion order then it not having index/slicing.
#### 2.When should we go for set data structure?
- If we want to represent a group of unique values as a single entity,then we should go for set.
- Set can't store duplicate elements.
- Insertition oder is not allowed.
##### Note:
- We can create set by using curly braces{} and all elements separated by comma separator in set.

#### 6.Creating set by using set(p) predefined function:
- We can create set by using set(p) function.
- Set(p) predefined function will take only one parameter.
- Thisshould be sequence (range,list,set,tuple,e.t.c....) object otherwise we will get error.
- Ex:

```python
r=range(0,10)
l=set(r)
print(l)
```

#### 9.Method in set data structure:
- As discussed,set is a predefined class.
- So,set class can contain methods because methods can be created inside of class only.
- We can check these methods by using dir(parameter1) predefined function.
- So, iternally set class contains two types of methods,
   * With underscore symbol method(no need)
   * Without  underscore symbol method(need)

```python
print(dir(set))
```

#### Important point:
- As per object-oriented principle,
   * If we want to access instance method then we should access by using object name.
   * So,all set methods we can access by using set object.
#### 10.methods:
    * i. add(p) method
    * ii. remove(p) method
    * iii. clear() method
##### i. add(p) method:
- add(p) is a method,we should access this method by using set object.
- This method adds elements to the set.

```python
s={45,65,25,15,35}
s.add(95)
print(s)
```

##### ii. remove(p) method:
- remove(p) is a method,we should access this method by using set object.
- This method removes elements from the set.
- If the specified element not present in the set,then we will get keyError.

```python
s={45,65,25,15,35}
s.remove(25)
s.remove(65)
print(s)
```

```python
# KeyError
s={45,65,25,15,35}
s.remove(115)
print(s)
```

##### iii. clear() method:
- clear() is a method,we should access this method by using set object.
- This method clear all elements from the set.

```python
s={45,65,25,15,35}
print(s)
s.clear()
print(s)
```

#### 11. Membership Operator:
- By using these operators ,we can find the specified element is exist in set or not.

```python
s={1,2,3,'ram'}

print(s)
print(2 in s)
print('a' in s)
print('q' in s)
print('m' not in s)
```

#### 12.Set comprehensions:
- Set comprehension  represents creating new set from iterable object like a list,set,tuple,dictionary and range.
- Set comprehension code is very concise way.

```python
s={x*x for x in range(5)}
print(s)
```

#### 13.Remove duplicates from list element:
- We can removes duplicates elements which are exists in list by passing list as parameter to set function.

```python
n=[1,2,3,4,1,1,2,3]
s=set(n)
print(s)
```
