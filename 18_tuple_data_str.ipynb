#### 1.Tuple data structure:
- We can create tuple data structure by using,
    * paranthesis() symbol
    * predefined tuple function
- A tuple can store group of object or elements.
    * A tuple can store same(homogeneous) type element.
    * A tuple can store different(heterogeneous) type of element.
- In tuple  inersertion oreder is preserved or fixed.
- If we insert element into 10,20,30 then o/p also will display as 10,20,30 then this is called as insertion order is preserved or fixed.
- Example:
    * input =>[10,20,30]
    * output =>[10,20,30]
- Duplicate elements are allowed.
- List having immutuable nature.
  * Immutual means once we create a list object then we can not change or modify the content of tuple object.
- Store elements by using index.
  * A tuple data structure supports both +ve and -ve index
  * +ve index means from left to right.
  * -ve index means from right to left.
#### Note:
- Inside tuple every object can be separated by comma(,) separator.
- tuple is a predifined class in python.
- Once id we create tuple object means inernally object is creating for tuple class.
- Table/Summary:
- 1. create -> ()paranthesis or tuple() function
  2. Store  -> Same data type or different data type
  3. insertion order ->fixed
  4. Duplicate ->allowed
  5. nature -> imMutuable
  6. index/slice ->-ve/+ve (both allowed)
#### 2.When should we go for tuple data structure:
- If we are going to deefine a data which never change overall the period,then we should go for the tuple data structure.

```python
emp_id=(10,20,30,40,50,60)
print(emp_id)
print(type(emp_id))
```

#### 5. Different ways to Creating a tuple:
###### (i)Empty tuple:
- We can create an empty tuple by using empty panathesis.

```python
emp_id=()
print(emp_id)
print(type(emp_id))
```

###### (ii)tuple with group of value:
- Tuple can contain grouo of object; those object can be same type or different type.

```python
emp_id=11,22,33,55,66,77
std_id=(11,22,33,55,66,77)
print(emp_id)
print(type(emp_id))
print(std_id)
print(type(std_id))
```

###### (iii) by using tuple(p) function:
- We can create tuple by using tuple(p) function.

```python
a=[11,22,33,44,55]
print(a)
print(type(a))
t=tuple(a)
print(t)
print(type(t))
```

#### 6.Accessing elements from tuple:
 - We can access from tuple by using,
    * i.index
    * ii.slice operator
    * iii.loops
##### i.index:
- Index means position where elements stores.


```python
t=(10,20,300,400,500)
print(t[0])
print(t[-1])
```

##### ii.slice operator:
- A group of object from starting point to ending point.

```python
t=(10,20,30,40,50)
print(t[2:5])
print(t[2::100])
print(t[::2])
print(t[:])
print(t[::])
```

##### iii.loop operator:

```python
value=(1,2,3,4,5,6,7,8)
for values in value:
    print(values[:])
```

#### 7.Tuple v/s immutability:
- Tuple having immutability nature.
- If we create a tuple then we can't modify the elements of existing tuple.

```python
t=(10,20,30,40,50)
print(t)
t[1]=70
print(t)
```

#### 8.Mathematical (+) & (*) operator on tuple:
- We can apply (+) and (*) operator on tuple.
   * i. (+) operator works as concatenation.
   * ii.(*) operator works as multiplication.
##### * i. (+) operator works as concatenation:
- (+) operator concatenates two tuples and returns single tuple.

```python
t=(10,20,30,40,50)
v=(60,70)
s=t+v
print(s)
```

##### ii.(*) operator works as multiplication:
- This operator works as repetition operator.

```python
t1=(10,20,30,40,50)
t2=t1*3
print(t2)
```

#### 11. Difference b/w list and tuple:
- List and tuple are exactly same except small difference.
   * list object are mutuable
   * tuple object are immutuable
- In both cases,
   * insertion order is preserved.
   * Duplicate objects are allowed.
   * Heterogenous objects are allowed.
   * Index and slicing are supported.
- Prove tuple is immutable by practice:

```python
values=(10,20,30,[40,50,60])
print(values)
print(values[0])
values[3][0]=99
print(values)
print(values[3][0])
print(values[3][1])
```
