## 2.Python program flow :
#### -> Now we will learn about the python program flow,it contains mainly three parts.
#### a.Source code , b.Byte code, c.Python virtual machine
## a.Source code :
#### ->The written python program is also called source code.
#### ->We need to save the program with either .py/.python extension.
####  EX-demo.py
#### ->Next step is , we need to run this progrsm by using py command.
#### EX-py demo.py
##  b.Byte code :
#### -> While runnning the program the interpreter convert the python program into Byte code.(Intermediate file)
#### ->This intermedfiate file is not available.
#### ->This file will be stored into "_py cache_folder" in current directory.
#### ->We can see this file by using below command.
####  EX-python_m py_compile demo .py
##  c.Python virtual machine :
#### ->This BYTE CODE file can't understand by the microprocessor to generate an output.
#### ->So,Python virtual machine convert this byte code into machine understandable format .
#### ->Finally we will get the output.
